## 5. Run Full Ablation Study

```python
from app.ml.deep_train import run_ablation_study
results=run_ablation_study(**config)
```

## 6. Results

```python
import pandas as pd
df=results['df_ablation']
print(df[['model','use_graph','use_weather','mae','rmse','coverage_p10_p90','winkler_score','beats_lightgbm']].to_string(index=False))
print()
print('Active default:',results['active_default_model'])
```

```python
import matplotlib.pyplot as plt
df_s=df.sort_values('mae')
cols=['#e74c3c' if m=='LightGBM_Conformal' else '#2ecc71' if b else '#3498db' for m,b in zip(df_s['model'],df_s['beats_lightgbm'])]
fig,axes=plt.subplots(1,2,figsize=(14,5))
axes[0].barh(df_s['model'],df_s['mae'],color=cols)
axes[0].set_xlabel('MAE (kW)'); axes[0].set_title('Test MAE by Model')
axes[0].axvline(df[df['model']=='LightGBM_Conformal']['mae'].values[0],color='red',linestyle='--',label='LightGBM')
axes[0].legend()
axes[1].barh(df_s['model'],df_s['winkler_score'],color=cols)
axes[1].set_xlabel('Winkler Score'); axes[1].set_title('Interval Quality')
plt.tight_layout()
plt.savefig('../results/ablation_chart.png',dpi=150,bbox_inches='tight')
plt.show()
```
